# Customer Churn Prediction Testing Notebook

This notebook sends prediction requests to the deployed model API and validates that the cloud deployment responds correctly using TF Serving JSON REST API.

```python
import json
import requests
import base64
import tensorflow as tf

# Update with your Back4App URL
API_URL = 'https://customerchurnapi-h387ilb8.b4a.run/v1/models/customer-churn-model:predict'

def _float_feature(value):
    return tf.train.Feature(float_list=tf.train.FloatList(value=[value]))

def _int64_feature(value):
    return tf.train.Feature(int64_list=tf.train.Int64List(value=[value]))

def _bytes_feature(value):
    return tf.train.Feature(bytes_list=tf.train.BytesList(value=[value.encode()]))

# Create a tf.train.Example
example = tf.train.Example(features=tf.train.Features(feature={
    'customer_age': _int64_feature(35),
    'gender': _bytes_feature('Male'),
    'contract_type': _bytes_feature('Month-to-Month'),
    'monthly_charges': _float_feature(72.5),
    'tenure': _int64_feature(12),
    'support_calls': _int64_feature(4),
    'total_usage': _float_feature(210.0),
    'satisfaction_score': _int64_feature(2),
}))

# Serialize and base64 encode
serialized_example = example.SerializeToString()
b64_example = base64.b64encode(serialized_example).decode('utf-8')

payload = {
    'instances': [{'b64': b64_example}]
}

response = requests.post(API_URL, json=payload, timeout=30)
print('Status Code:', response.status_code)
print('Prediction Response:', response.json())
```
